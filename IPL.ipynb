```python
# Loading the required libraries
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns 
```

```python
# Loading the ipl matches dataset
ipl=pd.read_csv('matches.csv')
```

```python
# Let's glance for the first five dataset records
ipl.head()
```

```python
# Looking in a number of rows and columns in ipl dataset
ipl.shape
```

```python
# Getting the frequency of most man of the match awards
ipl['player_of_match'].value_counts()
```

```python
# Getting the top 10 players with the most man of the match awards
ipl['player_of_match'].value_counts()[0:10]
```

```python
# Getting the top 5 players with the most man of the match awards
ipl['player_of_match'].value_counts()[0:5]
```

```python
# Getting the top 5 players with the most man of the match awards and ignores the count
list(ipl['player_of_match'].value_counts()[0:5].keys())
```

```python
# Display the top 5 players with the most match awards using a bar chart
plt.figure(figsize=(8,5))
plt.bar(list(ipl['player_of_match'].value_counts()[0:5].keys()),list(ipl['player_of_match'].value_counts()[0:5]), color='g')
plt.show()
```

```python
# Getting the frequency of result column
ipl['result'].value_counts()
```

```python
# Finding the number of toss_winners each time
ipl['toss_winner'].value_counts()
```

```python
# Extracting the records where a team won battling first
batting_first=ipl[ipl['win_by_runs']!=0]
```

```python
# Looking for the head
batting_first.head()
```

```python
# Making a histogram
plt.figure(figsize=(5,5))
plt.hist(batting_first['win_by_runs'])
plt.title("Distribution of Runs")
plt.xlabel("Runs")
plt.show()
```

```python
# Finding out the number of wins by each teanm after batting first
batting_first['winner'].value_counts()
```

```python
# Making bar plot for the top 3 teams with most wins after batting first
plt.figure(figsize=(6,6))

# Get the top 3 teams
top_teams = batting_first['winner'].value_counts()[0:3]

# Plot
plt.bar(top_teams.index.tolist(), top_teams.values.tolist(), color=["blue","yellow","orange"])
plt.title("Top 3 Teams with Most Wins After Batting First")
plt.xlabel("Team")
plt.ylabel("Number of Wins")
plt.show()
```

```python
# making a pie chart
import matplotlib.pyplot as plt
plt.figure(figsize=(7,7))
plt.pie(list(batting_first['winner'].value_counts()),labels=list(batting_first['winner'].value_counts().keys()),autopct='%0.1f%%')
plt.show()
```

```python
# Extracting those records a team won by batting second
batting_second=ipl[ipl['win_by_wickets']!=0]
```

```python
# Looking at the head
batting_second.head()
```

```python
#making histogram for frequency of wins w.r.t with wickets
plt.figure(figsize=(5,5))
plt.hist(batting_second['win_by_wickets'], bins=30)
plt.show()
```

```python
#finding out the frequency number of wins w.r.t each time after batting second
batting_second['winner'].value_counts()
```

```python
#Making a bar plot of the top 3 teams with the most wins after batting second
plt.figure(figsize=(5,5))
plt.bar(list(batting_second['winner'].value_counts()[0:3].keys()),list(batting_second['winner'].value_counts()[0:3]),color=["purple","blue","red"])
plt.show()
```

```python
#Making a pie chart distribution of most wins after batting second
plt.figure(figsize=(5,5))
plt.pie(list(batting_second['winner'].value_counts()),labels=list(batting_second['winner'].value_counts().keys()),autopct='%0.1f%%')
plt.show()
```
